# Итоговые выводы, изменения торгового алгоритма и дальнейший план

## 1. Итоги проверки исходных гипотез

### 1.1. Влияние рыночной фазы BTC

На исследуемой выборке сделки в `favorable` фазах показали более высокую результативность по сравнению с `unfavorable`: математическое ожидание составило 0.54R против 0.20R на сделку.

Таким образом, наблюдаемые результаты согласуются с исходной гипотезой о влиянии принятого разделения рыночных фаз BTC на эффективность стратегии. При этом статистический анализ показал высокую неопределённость оценки: доверительный интервал разницы оказался широким и включал нулевое значение. Поэтому величину и устойчивость эффекта пока нельзя считать надёжно установленными.

Несмотря на статистическую неопределённость, направление результата соответствует логике торговой системы: стратегия ориентирована на короткие позиции, поэтому спокойные фазы без выраженного доминирования покупателей являются более подходящей средой для её применения.

### 1.2. Эффективность повторных входов

Первичные и повторные входы показали практически одинаковый результат:

- первичные входы: математическое ожидание 0.43R;
- повторные входы: математическое ожидание 0.41R.

Наблюдаемая разница составляет всего −0.02R на сделку.

Исходная гипотеза заключалась не в том, что повторные входы должны превосходить первичные, а в том, что доступные повторные входы не должны уступать им настолько, чтобы их запрещать.

Для оценки была заранее установлена практически допустимая граница снижения математического ожидания −0.20R. Статистический анализ не позволил надёжно исключить более сильное ухудшение из-за малого количества повторных входов, однако фактически наблюдаемые торговые показатели двух групп практически совпали.

Полученный результат является основанием отказаться от полного запрета повторных входов и разрешить их применение при строгом соблюдении установленных условий.

### 1.3. Появление силы покупателя

Строгие признаки появления покупателя были зафиксированы только в 7 сделках, поэтому выборка недостаточна для статистически устойчивых выводов.

При этом даже на имеющихся наблюдениях обязательный досрочный выход при первом появлении данного признака ухудшил бы совокупный результат на 3.49R относительно фактического управления позициями.

Таким образом, текущие данные не дают оснований добавлять обязательный досрочный выход при появлении силы покупателя. Признак сохраняется для дальнейшего наблюдения, но изменение алгоритма на его основе не производится.


## 2. Основные закономерности, обнаруженные в ходе анализа

### 2.1. Отдельные рыночные фазы BTC

Декомпозиция агрегированных фаз показала, что результаты внутри групп `favorable` и `unfavorable` существенно неоднородны.

Наиболее выраженный результат получен в фазе `calm_decline`:

- 49 сделок;
- математическое ожидание 0.81R;
- профит-фактор 2.26;
- доля прибыльных сделок 38.8%.

Для сравнения, `calm_flat` при 37 наблюдениях показал математическое ожидание 0.20R.

Таким образом, преимущество агрегированной группы `favorable` в значительной степени связано именно с результатами `calm_decline`. Это одна из наиболее интересных закономерностей текущего исследования, поскольку она наблюдается не на единичных сделках, а на относительно крупной подвыборке.

При этом статистическая оценка разницы `calm_decline` относительно остальных фаз также имеет широкий доверительный интервал. Поэтому `calm_decline` следует рассматривать как сильного кандидата для дальнейшей проверки, а не как окончательно установленное преимущество.

### 2.2. Тип входа

Входы типа `spike` показали более высокую результативность по сравнению с `structure`:

- `spike`: математическое ожидание 0.81R;
- `structure`: математическое ожидание 0.26R.

Преимущество `spike` сохранялось как внутри `favorable`, так и внутри `unfavorable` фаз, поэтому наблюдаемая разница не объясняется только различиями в распределении рыночных фаз между двумя типами входа.

При этом `structure` также сохраняет положительное математическое ожидание и остаётся прибыльным типом входа. Полученный результат говорит не о необходимости отказаться от `structure`, а о различии относительной эффективности двух типов входа, которую необходимо будет проверить в будущей выборке.

### 2.3. Вид `spike`

Внутри входов `spike` сделки вида `stepped` показали значительно более высокий наблюдаемый результат по сравнению с `instant`:

- `stepped`: математическое ожидание 1.41R;
- `instant`: математическое ожидание 0.28R.

Однако каждая группа содержит около 20 наблюдений. Статистическая оценка показала очень широкий диапазон возможной величины эффекта, поэтому использовать этот результат как основание для изменения алгоритма пока рано.

Признак сохраняется для дальнейшего накопления данных.

### 2.4. Положение входа типа spike относительно уровня

Для входов `spike` сделки `above` показали значительно более высокий результат по сравнению с `below`:

- `above`: математическое ожидание 1.37R;
- `below`: математическое ожидание 0.12R.

При этом в текущей выборке преимущество above не объясняется заметным смещением распределения instant и stepped между группами. 

Однако размеры групп также невелики: 23 и 16 наблюдений соответственно. Поэтому найденная разница рассматривается как предварительная закономерность, требующая дальнейшего накопления данных.

### 2.5. Удалённость входа типа structure от уровня

Для входов `structure` сделки вблизи уровня показали более высокий наблюдаемый результат, чем удалённые входы:

- `close`: 72 сделки, математическое ожидание 0.31R;
- `far`: 18 сделок, математическое ожидание 0.04R.

Таким образом, преимущества более удалённого входа обнаружено не было. Напротив, текущая выборка указывает на возможное преимущество входов вблизи уровня.

Однако группа `far` содержит только 18 наблюдений, поэтому данный результат также требует дальнейшего накопления данных и не используется как основание для дополнительного запрета или фильтра.


## 3. Признаки, для которых полезных закономерностей не обнаружено

### 3.1. Время входа

Распределение сделок по часу входа не выявило естественных временных кластеров, которые можно было бы обоснованно использовать для разделения выборки.

Искусственное деление суток на произвольные интервалы не имеет достаточного логического основания с точки зрения текущей торговой системы. Поэтому дальнейшая сегментация сделок по времени входа не проводилась и дополнительный временной фильтр не вводится.

### 3.2. Числовые характеристики торговой ситуации

Не было обнаружено выраженной связи результата сделки со следующими числовыми признаками:

- spike_size;
- pullback_depth_m5;
- breakout_depth_m5;
- pullback_ratio.

Средние и медианные значения этих признаков в прибыльных и убыточных сделках оказались близки, а результаты сделок распределялись по их диапазонам без выраженной закономерности.

До начала анализа для этих признаков также не было сформулировано убедительного механизма, который объяснял бы их возможное влияние на результат с точки зрения поведения участников рынка.

Поэтому на текущем этапе дополнительная сегментация и дальнейшее накопление этих показателей не являются приоритетными. Они могут быть возвращены в анализ только при появлении новой предметной гипотезы.


## 4. Статистическая устойчивость результатов

Главным ограничением текущего исследования оказалась высокая неопределённость оценок.

Несмотря на заметные различия математического ожидания во многих группах, 95%-е доверительные интервалы всех проведённых групповых сравнений оказались широкими и включали нулевую разницу. Для повторных входов использовалась отдельная практически допустимая граница −0.20R, однако имеющаяся выборка также не позволила уверенно исключить её пересечение.

Причина заключается одновременно в ограниченном количестве наблюдений и в особенностях распределения торговых результатов. Большинство убыточных сделок находится около −1R, тогда как прибыльные сделки встречаются значительно реже и имеют средний размер +3,8R. Из-за этого наличие или отсутствие нескольких крупных прибыльных сделок существенно меняет среднее значение небольшой подвыборки.

Следовательно, высокое математическое ожидание отдельной группы ещё не означает, что его величина надёжно оценена.

Это не делает найденные закономерности бесполезными. Исследовательский анализ позволил определить направления, заслуживающие дальнейшего внимания, однако текущая выборка недостаточна для сильных статистических утверждений об истинной величине большинства обнаруженных эффектов.


## 5. Изменения торгового алгоритма

Результаты исследования используются для корректировки торговых правил. При этом изменения рассматриваются как рабочая версия алгоритма для следующего периода наблюдений, а не как окончательно доказанные оптимальные параметры.

### 5.1. Фильтр рыночной фазы

Для коротких позиций сохраняется торговля только в фазах, определённых как `favorable`.

Несмотря на широкую статистическую неопределённость, наблюдаемое направление эффекта, результаты декомпозиции отдельных фаз и логика самой стратегии дают достаточно оснований сохранить данный фильтр как рабочее правило.

Особое внимание в дальнейшем будет уделяться сравнению `calm_decline` и `calm_flat`, поскольку обе фазы остаются разрешёнными для торговли, но на текущей выборке показывают сильно различающиеся результаты.

### 5.2. Повторные входы

После первого неудачного входа разрешается одна повторная попытка как вида `spike`, так и вида `structure`, если она полностью соответствует установленным условиям торговой системы.

При этом увеличение допустимого количества повторных входов не должно приводить к снижению требований к качеству сигнала. После первого убытка особенно важно не искать повторный вход на эмоциях и не интерпретировать обычное движение рынка как новый сигнал только из-за желания компенсировать предыдущую потерю.

Таким образом, повторный вход рассматривается не как обязательное продолжение первой сделки, а как отдельная новая возможность, которая должна самостоятельно соответствовать всем правилам.

### 5.3. Дневной риск

В связи с разрешением повторных входов максимальное количество потенциальных попыток увеличивается до двух первичных и двух повторных входов в течение торгового дня.

Предельный дневной риск повышается с 3R до 4R.


## 6. План дальнейшего сбора данных

Следующий этап исследования будет строиться не вокруг повторного поиска закономерностей в той же выборке, а вокруг проверки уже сформулированных выводов на новых сделках.

Правила и основные гипотезы фиксируются сейчас, после чего будет собираться новая выборка сделок, которые ещё не существовали на момент проведения текущего анализа. Это позволит проверить, сохраняются ли обнаруженные закономерности на новых данных, а не являются ли они особенностью конкретного исторического периода.

В первую очередь продолжается сбор наблюдений по следующим признакам:

- btc_phase: `calm_decline` относительно `calm_flat`;
- entry_type: `spike` относительно `structure`;
- spike_type: `stepped` относительно `instant`;
- entry_vs_level (spike): `above` относительно `below`;
- entry_vs_level (structure): `close` относительно `far`;
- эффективность повторных входов.

Числовые признаки, которые не показали закономерности и для которых отсутствует убедительная предметная гипотеза, не являются приоритетом дальнейшего исследования.

Поскольку после введения фазового фильтра реальные короткие сделки в `unfavorable` фазах совершаться не будут, полноценное повторное сравнение `favorable` и `unfavorable` на фактически совершённых сделках станет невозможным. Если в дальнейшем потребуется отдельно проверить качество самого фазового фильтра, можно фиксировать валидные сигналы в исключённых фазах без реального открытия позиции и рассчитывать их результат по заранее установленным правилам.

По мере накопления новой выборки могут добавляться новые признаки и гипотезы, но они должны формулироваться до анализа их результатов, а не подбираться после просмотра данных.

Отдельным направлением дальнейшей работы станет разработка правил для длинных позиций. Для них потребуется самостоятельное определение подходящих рыночных фаз и критериев входа. Расчёт корреляции с BTC может сохранять ту же общую логику, однако длинные и короткие сделки должны анализироваться раздельно. Текущий проект относится только к существующей системе коротких позиций.


## 7. Ограничения исследования

При интерпретации результатов необходимо учитывать ряд ограничений.

1. **Общий размер выборки.** В анализ вошло 130 сделок. Этого достаточно для первичного исследования стратегии, но недостаточно для точной оценки большинства различий между подгруппами.

2. **Малые размеры отдельных групп.** Для некоторых наиболее интересных закономерностей количество наблюдений составляет около 15–25 сделок на группу. Поэтому даже крупные наблюдаемые различия имеют высокую статистическую неопределённость.

3. **Асимметричное распределение результатов.** Большинство убыточных сделок имеет результат около −1R, а положительный итог стратегии формируется меньшим количеством значительно более крупных прибыльных сделок. Это увеличивает изменчивость среднего результата и ширину доверительных интервалов.

4. **Частичная зависимость сделок.** Отдельные сделки нельзя считать полностью независимыми друг от друга. Несколько сделок могут совершаться в один день и находиться под влиянием одного и того же движения рынка.

5. **Влияние продолжительных рыночных режимов.** Сделки, совершённые в близкие даты, могут находиться внутри одной долгосрочной фазы рынка. Поэтому десятки формально отдельных наблюдений частично отражают одни и те же внешние условия.

6. **Неравномерное распределение рыночных фаз во времени.** Разные части исследуемого периода содержат разное количество фаз каждого типа, что затрудняет полное отделение эффекта торгового правила от особенностей конкретного рыночного периода.

7. **Множественный поиск закономерностей.** Помимо исходных гипотез в ходе исследовательского анализа рассматривалось несколько дополнительных признаков. Чем больше сравнений проводится на одной выборке, тем выше вероятность обнаружить случайно возникшее различие. Поэтому найденные дополнительные закономерности требуют проверки на новых данных.

8. **Часть признаков определяется вручную.** Классификация рыночных фаз, вида входа и некоторых особенностей торговой ситуации основана на формализованных, но частично визуальных критериях. Это создаёт возможность погрешности классификации.

9. **Ограниченная переносимость результатов.** Полученные выводы относятся к конкретной системе коротких позиций, используемым фильтрам, правилам управления позицией и исследованному периоду. Они не должны автоматически переноситься на длинные позиции, другие рынки или существенно изменённый алгоритм.


## 8. Итог

Исследование позволило формализовать торговую систему, отделить действующие правила от исторических решений, количественно оценить её общие показатели, проверить исходные гипотезы и выявить несколько новых потенциально важных закономерностей.

Наиболее практически значимыми результатами стали введение фазового фильтра и увеличение частоты повторных входов.

Одновременно статистический анализ показал, что текущего количества наблюдений недостаточно для точной оценки большинства найденных различий. Поэтому более детальные закономерности — преимущество `spike`, `calm_decline`, `stepped`, `above` и возможное преимущество `close` — пока не превращаются в дополнительные жёсткие фильтры.

Следующая выборка должна использоваться прежде всего для проверки устойчивости уже сформулированных закономерностей. Основная задача следующего этапа — не найти как можно больше новых различий, а определить, какие из результатов текущего исследования воспроизводятся на новых данных.

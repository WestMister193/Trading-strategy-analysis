# Trading Strategy Analysis

Данный проект является анализом стратегии торговли на криптовалютном рынке. 

Суть стратегии: открытие коротких позиций по инструментам, соответствующим условиям: низкая корреляция с BTC, выраженная сила продавца, понятный уровень поддержки, достаточный потенциал хода, а также условия для выбранного типа входа.

## Обзор исследования

### Цель исследования: 
Cформулировать обоснованные рекомендации по улучшению текущей стратегии и выявить новые закономерности для проверки на дальнейших выборках

### Задачи:
1. Сформировать датасет сделок в виде excel-файла, расписать сделки по выбранным метрикам и предобработать данные для анализа
2. Выполнить исследовательский анализ, включающий в себя одномерный анализ метрик, проверку предварительно сформулированных гипотез, а также сравнительный и комплексный анализ метрик с целью поиска новых закономерностей
3. Провести статистический анализ результатов, полученных в ходе исследования, для оценки надежности выявленных различий и неопределенности полученных оценок
4. На основе полученных наблюдений сформулировать выводы, рекомендации и новые гипотезы.
5. Создать интерактивный дашборд для визуализации данных и полученных результатов

### Инструменты: 
- Excel для формирования датасета
- JupiterLab с использованием языка программирования Python для исследовательского и статистического анализов
- Power BI для визуализации

### Предварительно сформулированные гипотезы: 

H1. Влияние рыночной фазы BTC.
Сделки в фазах BTC calm_flat и calm_decline должны показывать более высокую результативность, чем сделки в остальных фазах.

H2. Эффективность повторного входа.
Повторные открытия сделок после получения убытка не показывают значительно худших результатов по сравнению с первичными входами.

H3. Выход при появлении сильного покупателя.
Досрочное закрытие позиции при появлении заранее определённых признаков силы покупателя может улучшать результат управления позицией.

## Данные и методология

Анализ проводится на основании датасета из 130 сделок за период 03.01.2026 - 13.09.2026. Все отобранные сделки соответствуют актуальному алгоритму торговли.
В рамках анализа результаты сделок приведены к единицам риска R, а не сравниваются напрямую в долларах США. Отклонение фактического результата убыточной сделки от расчётного −1R связано с влиянием переменных торговых издержек.
В датасете представлены как категориальные, так и числовые признаки для описания сделок. Также часть признаков относится к сделкам только одного из типов входа. Подробное описание всех признаков представлено в [Data Dictionary](data/data_dictionary.xlsx).

## Исследовательский анализ

На первом этапе был проведён общий анализ результатов стратегии и распределения отдельных признаков, после чего проверены предварительно сформулированные гипотезы и выполнен поиск дополнительных закономерностей.

Общий результат по 130 сделкам:

| Метрика | Результат |
| --- | ---: |
| Количество сделок | 130 |
| Суммарный результат | +55.77R |
| Средний результат сделки | +0.43R |
| Profit Factor | 1.58 |
| Доля прибыльных сделок | 30.8% |

### Проверка предварительных гипотез

**H1. Влияние рыночной фазы BTC**

В предварительно определённых благоприятных фазах BTC было совершено 86 сделок со средним результатом +0.54R.  
В остальных фазах — 44 сделки со средним результатом +0.20R.

Наиболее высокий результат среди отдельных фаз наблюдался в `calm_decline`: 49 сделок со средним результатом +0.81R.

Таким образом, направление исходной гипотезы подтверждается на уровне наблюдаемой выборки, однако надежность различий дополнительно оценивается в статистическом анализе.

**H2. Эффективность повторного входа**

Первичные входы:
- 112 сделок;
- средний результат +0.43R;
- Profit Factor 1.59.

Повторные входы:
- 18 сделок;
- средний результат +0.41R;
- Profit Factor 1.52.

На наблюдаемой выборке повторные входы не показали заметного ухудшения результатов относительно первичных.

Из 77 убыточных первичных входов в 35 случаях появилась релевантная возможность повторного входа. В 18 случаях повторный вход был фактически реализован.

**H3. Выход при появлении сильного покупателя**

Признаки сильного покупателя были зафиксированы только в 7 сделках.

Контрфактическое сравнение показало, что обязательное закрытие позиции в момент появления этого признака уменьшило бы суммарный результат этих сделок на 3.49R.

Полученные данные не дают оснований добавлять обязательный досрочный выход по данному признаку.

### Дополнительные наблюдения

В ходе исследовательского анализа были обнаружены несколько потенциально интересных закономерностей:

- сделки типа `spike` показали средний результат +0.81R против +0.26R для `structure`;
- внутри `spike` сделки типа `stepped` показали более высокий наблюдаемый результат, чем `instant`;
- входы `above` показали более высокий наблюдаемый результат, чем `below`;
- особенно высокая результативность наблюдалась в фазе `calm_decline`.

В то же время выраженных различий между прибыльными и убыточными сделками по исследованным числовым признакам размера spike, глубины пробоя и глубины отката обнаружено не было.

Обнаруженные в ходе исследовательского анализа различия рассматриваются как кандидаты для дальнейшей проверки, а не как подтверждённые закономерности.

Полный исследовательский анализ представлен в [02_eda.ipynb](notebooks/02_eda.ipynb).

## Статистический анализ

Для оценки надежности обнаруженных различий использовался непараметрический bootstrap разницы средних результатов сделки. Для каждого сравнения выполнялось 10 000 повторных выборок с возвращением.

Основные результаты:

| Сравнение | Наблюдаемая разница | 95% доверительный интервал |
| --- | ---: | ---: |
| favorable vs unfavorable | +0.34R | [-0.49R; +1.16R] |
| spike vs structure | +0.55R | [-0.33R; +1.46R] |
| above vs below | +1.24R | [-0.26R; +2.69R] |
| stepped vs instant | +1.13R | [-0.43R; +2.61R] |
| calm_decline vs остальные фазы | +0.60R | [-0.21R; +1.43R] |

Во всех приведённых сравнениях доверительный интервал включает ноль. Поэтому текущая выборка не позволяет считать наблюдаемые различия статистически устойчивыми.

Для повторных входов отдельно проверялась гипотеза о том, что их результативность не хуже первичных более чем на практически допустимую величину. Наблюдаемые средние результаты были практически одинаковыми, однако доверительный интервал оказался достаточно широким и пересекал выбранную границу не худшей результативности −0.2R. Поэтому статистически подтвердить эту гипотезу на текущем объёме данных не удалось.

H3 статистически отдельно не проверялась из-за малого количества наблюдений — всего 7 случаев появления сильного покупателя.

Таким образом, статистический анализ не опровергает найденные в EDA закономерности, но показывает высокую неопределённость их текущих оценок. Для более уверенных выводов требуется дальнейшее накопление данных.

Полный статистический анализ представлен в [03_statistical_analysis.ipynb](notebooks/03_statistical_analysis.ipynb).

## Выводы и рекомендации

По результатам исследования были сформулированы следующие решения для дальнейшего применения и проверки стратегии:

1. Использовать для коротких сделок предварительно определённые благоприятные фазы BTC — `calm_flat` и `calm_decline`. На текущей выборке они показывают более высокий результат, однако эффект требует дальнейшего подтверждения на новых данных.

2. Не запрещать релевантные повторные входы после первой убыточной попытки. Их наблюдаемая результативность близка к результативности первичных входов. При этом критерии повторного входа не должны смягчаться после получения убытка.

3. Не вводить обязательный досрочный выход при появлении признаков сильного покупателя. На имеющейся небольшой выборке такой вариант управления позицией ухудшил бы результат.

4. Не изменять размер риска в зависимости от типа входа `spike` / `structure`. Несмотря на заметную разницу результатов в текущей выборке, статистическая неопределённость остаётся высокой.

5. Продолжить сбор данных для проверки выявленных в ходе EDA закономерностей, в первую очередь:
   - преимущества `calm_decline`;
   - различия между `spike` и `structure`;
   - различия между `stepped` и `instant`;
   - различия между `above` и `below`.

Следующая выборка должна использоваться прежде всего для проверки уже сформулированных на текущих данных гипотез, а не для повторного подбора правил под имеющуюся историю.

Более подробные результаты исследования, рекомендации, направления дальнейшей проверки и ограничения анализа представлены в [итоговых выводах](notebooks/04_conclusions.ipynb).
